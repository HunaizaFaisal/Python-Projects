```python
## Task 6: Grocery Shopping Bill
### Task Description
#Customer ke purchased products ka complete grocery bill calculate karein.
### Requirements
#1. User se total products input lein.
#2. Har product ka name, price aur quantity input lein.
#3. Har product ka subtotal calculate karein.
#Subtotal = Price × Quantity
#1. Tamam subtotals ko add karke total bill calculate karein.
#2. Rs. 5,000 ya above ke bill par 10% discount dein.
#3. Final bill display karein.
total_bill=0
i = int(input("Enter Number of Products"))
for n in range (i) :
  products_name=input("Enter Product Name")
  price=int(input("Enter Price"))
  quantity=int(input("Enter Quantity"))
  subtotal=price*quantity
  total_bill=total_bill + subtotal
if total_bill >= 5000:
  discount=total_bill*10/100
  final_bill=total_bill-discount
  print(final_bill)
else:
  print("Total Bill",total_bill)
```

```python
## Task 7: Employee Weekly Salary
### Task Description
#Employee ki weekly working hours ke according salary calculate karein.
### Rules
#- Normal working limit: 40 hours
#- First 40 hours: Normal hourly rate
#- Extra hours: 1.5 times hourly rate
### Requirements
#1. Employee name input lein.
#2. Total working hours input lein.
#3. Hourly rate input lein.
#4. Normal salary calculate karein.
#5. Overtime salary calculate karein.
#6. Total weekly salary display karein.
employee_name=input("Enter Employee Name")
total_working_hours=int(input("Enter Total Working Hours"))
hourly_rate=int(input("Enter Hourly Rate"))
if total_working_hours <= 40:
  normal_salary=total_working_hours*hourly_rate
  print(normal_salary)
elif total_working_hours > 40:
  overtime_salary=(total_working_hours-40)*1.5*hourly_rate
  normal_salary=40*hourly_rate
  total_salary=normal_salary+overtime_salary
  print("Total Working Hours",total_working_hours)
  print("Total overtime Salary",overtime_salary)
  print("Total Salary",total_salary)
```

```python
## Task 6: Grocery Shopping Bill
### Task Description
#Customer ke purchased products ka complete grocery bill calculate karein.
### Requirements
#1. User se total products input lein.
#2. Har product ka name, price aur quantity input lein.
#3. Har product ka subtotal calculate karein.
#Subtotal = Price × Quantity
#1. Tamam subtotals ko add karke total bill calculate karein.
#2. Rs. 5,000 ya above ke bill par 10% discount dein.
#3. Final bill display karein.
total_bill=0
i = int(input("Enter Number of Products"))
for n in range (i) :
  products_name=input("Enter Product Name")
  price=int(input("Enter Price"))
  quantity=int(input("Enter Quantity"))
  subtotal=price*quantity
  total_bill=total_bill + subtotal
if total_bill >= 5000:
  discount=total_bill*10/100
  final_bill=total_bill-discount
  print(final_bill)
else:
  print("Total Bill",total_bill)
```

```python
## Task 9: Message Analyzer
# Task Description
#User se ek message input lein aur message ke baare mein different information display karein.
### Requirements
#Program calculate kare:
#1. Total characters
#2. Total words
#3. Total vowels
#4. Total spaces
#5. Message uppercase format
#6. Check karein ke word `"Python"` message mein exists karta hai ya nahi.
message = input("Enter Message")
# Total Characters
print(f"The total characters in the message is {len(message)}")
# Total Words
print(f"The total words in the message is {len(message.split())}")
# Total Vowels
vowels = "aeiouAEIOU"
total_vowels = 0

for char in message:
    if char in vowels:
        total_vowels += 1
print(f"The total vowels in the message is {total_vowels}")
#Total Spaces
print(f"The total spaces in the message is {message.count(' ')}")
#Message Uppercase Format
print(f"The message in uppercase is {message.upper()}")
# word python
if "python" in message or "Python" in message:
  print("The word python exist in message")
else:
  print("No it doesn't")
```

```python
#Create this data:

#Months = Jan, Feb, Mar, Apr, May
#Sales  = 20, 35, 30, 50, 45
#Create a **line chart**.
#Add:
#- Title- X-axis label- Y-axis label
import matplotlib.pyplot as plt
import pandas as pd
months=[ "Jan","Feb","Mar","Apr","May"]
sales=[20,35,30,50,45]
df = pd.DataFrame({"Months": months, "Sales": sales})
df
plt.bar(df["Months"], df["Sales"])
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
#Create a scatter plot using:
#Study Hours = 1, 2, 3, 4, 5, 6
#Marks       = 40, 50, 60, 70, 82, 90
import matplotlib.pyplot as plt
import pandas as pd
study_hours=[1,2,3,4,5,6]
marks=[40,50,60,70,82,90]
df = pd.DataFrame({"Study Hours": study_hours, "Marks": marks})
df
plt.scatter(df["Study Hours"], df["Marks"])
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
#Create a histogram using these marks:
#45, 50, 55, 60, 60, 65, 70, 70, 75, 80, 85, 90
import matplotlib.pyplot as plt
import pandas as pd
marks=[45,50,55,60,60,65,70,70,75,80,85,90]
df = pd.DataFrame({"Marks": marks})
df
plt.hist(df["Marks"])
plt.title("Marks Histogram")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```
